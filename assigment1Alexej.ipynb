```python
from bs4  import BeautifulSoup as bp
import requests as req
import json
import pandas as pd

url = "https://www.zdfheute.de/politik/deutschland/weidel-afd-spenden-affaere-leaks-100.html"
response = req.get(url)
htmlDoc = response.text
soup = bp(htmlDoc, 'html.parser')

# Vars to be extracted
title = ''
authors = []
publicationDate = ''
mainText = ''

# in <head>...<title>xxxx</title>...</head>
title = soup.find('title').text

script_tags = soup.find_all(
    "script",
    type="application/ld+json"
)

jsonData = {}

for tag in script_tags:
    if '"author"' in tag.text:
        jsonData = json.loads(tag.text)
        break
        
# print(jsonData.keys()) 
# print(jsonData.values())

for author in jsonData.get('author'):
    authors.append(author.get('name'))

publicationDate = jsonData.get('datePublished')

for elem in soup.find_all('p'):
    mainText += elem.text + "\n"

# df = pd.DataFrame({
#     "Publication Timestamp": [publicationDate],
#     "Author": [authors],
#     "Title": [title],
#     "Main Text": [mainText]
# })

#display(df)

print('[Publication Timestamp]')
print(publicationDate)
print('[authors]')
print(authors)
print('[title]')
print(title)
print('[Main Text]')
print(mainText)
```
